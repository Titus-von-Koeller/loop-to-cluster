```python
"""fp32 single-GPU baseline: SmolLM2-135M trained from scratch on wikitext-2."""

import math

import torch
from datasets import load_dataset
from torch.utils.data import DataLoader
from transformers import AutoConfig, AutoModelForCausalLM, AutoTokenizer

model_name = "HuggingFaceTB/SmolLM2-135M"
device = "cuda:0"
num_epochs = 3
batch_size = 8
seq_len = 128
learning_rate = 1e-4

torch.manual_seed(0)

cfg = AutoConfig.from_pretrained(model_name)
cfg.dtype = torch.float32  # override model's default dtype of bfloat16 for baseline
cfg.use_cache = False  # disable caching, as it's only useful for inference

model = AutoModelForCausalLM.from_config(cfg)
model.to(device)
model.train()
model
```
